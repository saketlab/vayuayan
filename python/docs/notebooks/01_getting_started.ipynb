# Start with vayuayan


## 1. Import the package


```python
import json

import pandas as pd

from vayuayan import CPCBHistorical, CPCBLive
```

## 2. Find the available data


```python
aqi_client = CPCBHistorical()

states = aqi_client.get_state_list()
print(f"Total states available: {len(states)}")
print("\nFirst 10 states:")
for state in states[:10]:
    print(f"  - {state}")
```

### Get the cities in a state


```python
cities = aqi_client.get_city_list("Maharashtra")
print(f"Cities in Maharashtra: {len(cities)}")
print("\nCities:")
for city in cities:
    print(f"  - {city}")
```

### Get the monitoring stations in a city


```python
stations = aqi_client.get_station_list("Mumbai")
print(f"Monitoring stations in Mumbai: {len(stations)}")
print("\nStations:")
for station in stations:
    print(f"  - {station['value']}: {station['label']}")
```

## 3. Get live AQI data


```python
live_client = CPCBLive()

lat, lon = live_client.get_system_location()
# IP geolocation is approximate, so 1 decimal (about 11 km) is enough
print(f"Your approximate location: {lat:.1f}, {lon:.1f}")
```

### Find the nearest monitoring station


```python
nearest_station = live_client.get_nearest_station()
print("Nearest monitoring station:")
print(json.dumps(nearest_station, indent=2))
```

### Get the current AQI data


```python
current_aqi = live_client.get_live_aqi_data()
current_aqi.pop("last_hours", None)
print("Current Air Quality Data:")
print(json.dumps(current_aqi, indent=2))
```

## 4. Download historical data


```python
output_file = "mumbai_aqi_2024.csv"
aqi_client.download_past_year_aqi_data_city_level("Mumbai", "2024", output_file)
print(f"Data downloaded to {output_file}")

df = pd.read_csv(output_file)
print(f"\nDataset shape: {df.shape}")
print("\nFirst few rows:")
print(df.head())
```

```python
# A station ID (site_5104) or a part of the station name gives the same
# station.

output_file = "kurla_2024.csv"
aqi_client.download_past_year_aqi_data_station_level(
    "Kurla, Mumbai", "2024", output_file
)
print(f"Data downloaded to {output_file}")

df = pd.read_csv(output_file)
print(f"\nDataset shape: {df.shape}")
print("\nFirst few rows:")
print(df.head())
```
