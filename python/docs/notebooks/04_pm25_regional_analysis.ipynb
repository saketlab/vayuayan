# PM2.5 Regional Analysis


```python
import geopandas as gpd
import matplotlib.pyplot as plt

from vayuayan import WHO_PM25_GUIDELINES, PM25Client, pollutant_category
```

## 1. About the PM2.5 Data


## 2. Initialize the PM2.5 Client


```python
# Initialize the PM2.5 client
pm25_client = PM25Client()

print("PM2.5 Client initialized successfully")
```

## 3. Download the GeoJSON File of the Indian States


```python
# Download and save the GeoJSON file of the Indian states

sample_geojson = gpd.read_file(
    "https://gist.githubusercontent.com/JaggeryArray/fa31964eedb0c2da023c9485772f911a/raw/02c0644de34fbae9dbac2ba0496a00772a2c28cd/india_map_states.geojson"
)

# Save to file
geojson_file = "indian_states.geojson"
sample_geojson.to_file(geojson_file, driver="GeoJSON")
```

## 4. Show the GeoJSON Region on a Map


```python
# Load and visualize the GeoJSON
gdf = gpd.read_file(geojson_file)

fig, ax = plt.subplots(figsize=(5, 5))
gdf.plot(ax=ax, facecolor="lightblue", edgecolor="black", linewidth=2)
ax.set_title("Analysis Region", fontsize=16, fontweight="bold")
ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("\nRegion Details:")
print(f"CRS: {gdf.crs}")
print(f"Number of Features: {len(gdf)}")
print(f"Columns: {gdf.columns.tolist()}")
print(f"Name: {gdf['state_name'].iloc[0]}")
print(f"Bounds: {gdf.total_bounds}")
```

## 5. Get the PM2.5 Statistics for All of India in 2023


```python
# Get PM2.5 statistics for the region
# Give the year and, if necessary, the month

year = 2023
month = 2  # February (optional. Use None for annual data.)

stats = pm25_client.get_pm25_stats(geojson_file, year, month)

print(f"PM2.5 Statistics for {year}/{month if month else 'Annual'}:")
print(f"\nMean PM2.5: {stats['mean']:.2f} μg/m³")
print(f"Standard Deviation: {stats['std']:.2f} μg/m³")
print(f"Minimum PM2.5: {stats['min']:.2f} μg/m³")
print(f"Maximum PM2.5: {stats['max']:.2f} μg/m³")

# The Indian AQI category of the mean, from the CPCB PM2.5 breakpoints. They
# are for 24-hour averages, so for a monthly mean this is an indication only.
print(f"\nAQI category of the mean: {pollutant_category('PM2.5', stats['mean'])}")
times = stats["mean"] / WHO_PM25_GUIDELINES["annual"]
print(
    f"{times:.1f} times the WHO annual guideline ({WHO_PM25_GUIDELINES['annual']} μg/m³)"
)
```

## 6. Analyze the Sub-Regions


```python
# Get PM2.5 statistics for each state, with the state polygons for a map
df_results = pm25_client.get_pm25_stats(
    geojson_file, year=2023, month=2, group_by="state_name", geometry=True
)

print("PM2.5 Statistics by State:")
print(df_results[["state_name", "mean", "std", "min", "max", "count"]])

# Find the state with the highest pollution
worst_state = df_results.loc[df_results["mean"].idxmax()]
print(f"\nMost Polluted State: {worst_state['state_name']}")
print(f"Average PM2.5: {worst_state['mean']:.2f} μg/m³")
```

## 7. Show the PM2.5 Distribution


```python
# df_results is a GeoDataFrame, so it plots as a choropleth map directly
fig, ax = plt.subplots(figsize=(12, 10))
df_results.plot(
    column="mean",
    ax=ax,
    legend=True,
    cmap="YlOrRd",
    edgecolor="black",
    linewidth=1.5,
)

# Add state labels
for _, row in df_results.iterrows():
    centroid = row.geometry.centroid
    ax.text(
        centroid.x,
        centroid.y,
        f"{row['state_name']}\n{row['mean']:.1f}",
        ha="center",
        va="center",
        fontsize=8,
        bbox=dict(boxstyle="round", facecolor="white", alpha=0.7),
    )

ax.set_title(
    "PM2.5 Distribution by State (μg/m³), February 2023", fontsize=16, fontweight="bold"
)
ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")
plt.tight_layout()
plt.show()
```

## 8. Analysis over Time


```python
# PM2.5 statistics for each month of 2023 (months that cannot be downloaded are left out)
df_monthly = pm25_client.get_pm25_monthly_stats(geojson_file, 2023)
print(df_monthly[["month", "mean", "std"]].round(2))

if not df_monthly.empty:
    plt.figure(figsize=(14, 6))
    plt.plot(
        df_monthly["month"], df_monthly["mean"], marker="o", linewidth=2, markersize=8
    )
    plt.fill_between(
        df_monthly["month"],
        df_monthly["mean"] - df_monthly["std"],
        df_monthly["mean"] + df_monthly["std"],
        alpha=0.3,
    )
    plt.xlabel("Month", fontsize=12)
    plt.ylabel("PM2.5 (μg/m³)", fontsize=12)
    plt.title("Monthly PM2.5 Trends, 2023", fontsize=14, fontweight="bold")
    plt.grid(True, alpha=0.3)
    plt.xticks(range(1, 13))
    plt.tight_layout()
    plt.show()

    worst_month = df_monthly.loc[df_monthly["mean"].idxmax()]
    best_month = df_monthly.loc[df_monthly["mean"].idxmin()]
    print(
        f"\nWorst Month: {int(worst_month['month'])} ({worst_month['mean']:.2f} μg/m³)"
    )
    print(f"Best Month: {int(best_month['month'])} ({best_month['mean']:.2f} μg/m³)")
```

## 9. Export the Results


```python
# Export results to CSV
df_results.drop(columns="geometry").to_csv("pm25_analysis_results.csv", index=False)
print("State-level results saved to: pm25_analysis_results.csv")

df_monthly.to_csv("pm25_monthly_trends.csv", index=False)
print("Monthly trends saved to: pm25_monthly_trends.csv")
```
